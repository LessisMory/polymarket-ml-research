```python
import pandas as pd

# 1. 读取两张总表
df_ob = pd.read_parquet('../data/OB_UP_DATA.parquet')
df_hft = pd.read_parquet('../data/grand_total_HFT_master.parquet')

# ==========================================
# 🚨 关键修复：强制统一时区为 UTC
# ==========================================
df_ob['timestamp'] = pd.to_datetime(df_ob['timestamp'], utc=True)
df_hft['timestamp'] = pd.to_datetime(df_hft['timestamp'], utc=True)

# 2. 按时间和合约名称合并
df_merged = pd.merge(df_ob, df_hft, on=['timestamp', 'slug'], how='left')

# 检查合并结果
print("合并成功！数据维度：", df_merged.shape)
```

```python
df_merged.to_parquet('../data/ultimate_data.parquet', index=False)
```
